**Attribute Information**

This dataset contains 24 columns and 3000 rows.

- **Date**: Date on which the data was recorded
- **Stock Index**: Name of the stock market index (Dow Jones, S&P 500, NASDAQ)
- **Open Price**: Opening price of the stock index
- **Close Price**: Closing price of the stock index
- **Daily High**: Highest price of the index during the day
- **Daily Low**: Lowest price of the index during the day
- **Trading Volume**: Total number of shares traded
- **GDP Growth (%)**: Growth rate of the economy
- **Inflation Rate (%)**: Percentage increase in prices
- **Unemployment Rate (%)**: Percentage of unemployed labor force
- **Interest Rate (%)**: Central bank interest rate
- **Consumer Confidence Index**: Measure of consumer optimism
- **Government Debt (Billion USD)**: Total government debt
- **Corporate Profits (Billion USD)**: Profits earned by corporations
- **Forex USD/EUR**: Exchange rate between USD and EUR
- **Forex USD/JPY**: Exchange rate between USD and JPY
- **Crude Oil Price (USD per Barrel)**: Price of crude oil
- **Gold Price (USD per Ounce)**: Price of gold
- **Real Estate Index**: Index representing real estate market performance
- **Retail Sales (Billion USD)**: Total retail sales value
- **Bankruptcy Rate (%)**: Percentage of bankruptcies
- **Mergers & Acquisitions Deals**: Number of M&A deals
- **Venture Capital Funding (Billion USD)**: Amount of venture capital investment
- **Consumer Spending (Billion USD)**: Total consumer expenditure

<p style="font-size:20px;"><b>Step 2:Data Pre-processing</b><br>

<p style="font-size:15px;">In this step, the dataset is cleaned and prepared for analysis.

```python
#Checking missing values.
missing= df.isnull().sum()
missing[missing>0]
```

```python
#Removing duplicates 
duplicates = df.duplicated().sum()
duplicates
```

**Correcting data types**

```python
# Converting Date column
df['Date']= pd.to_datetime(df['Date'],format="%d-%m-%Y")
```

```python
# Identifying Numeric columns
numeric_cols= ['Open Price', 'Close Price', 'Daily High', 'Daily Low', 'Trading Volume',
    'GDP Growth (%)', 'Inflation Rate (%)', 'Unemployment Rate (%)',
    'Interest Rate (%)', 'Consumer Confidence Index',
    'Government Debt (Billion USD)', 'Corporate Profits (Billion USD)',
    'Forex USD/EUR', 'Forex USD/JPY', 'Crude Oil Price (USD per Barrel)',
    'Gold Price (USD per Ounce)', 'Real Estate Index',
    'Retail Sales (Billion USD)', 'Bankruptcy Rate (%)',
    'Mergers & Acquisitions Deals', 'Venture Capital Funding (Billion USD)',
    'Consumer Spending (Billion USD)']
```

```python
#Cleaning Numeric values
df[numeric_cols]=df[numeric_cols].replace({',':''},regex=True).astype(float)
```

```python
#Converting all Numeric values to float
df[numeric_cols]= df[numeric_cols].apply(pd.to_numeric,errors='coerce')
```

```python
#Keeping stock Index as String
df['Stock Index']= df['Stock Index'].astype(str)
```

```python
#Checking data types
df.dtypes
```

**Creating derived columns**
<p style="font-size:15px;">Derived variables such as daily stock returns, volatility, and composite economic indices were created 
to better capture financial performance and macroeconomic conditions.
</p>

```python
#Daily Stock Return(%)-Shows daily market performance.
df['Daily Stock Return(%)']= ((df['Close Price']- df['Open Price'])/df['Open Price'])*100
```

```python
df[['Open Price', 'Close Price','Daily Stock Return(%)']].head()
```

```python
#Daily Volatility (%)- Measures market risk for the day.
df['Daily Volatility(%)']= ((df['Daily High']-df['Daily Low'])/df['Open Price'])*100
```

```python
df['Daily Volatility(%)'].head()
```

```python
#Economic Pressure Index (Composite Indicator)-It combines multiple economic stress factors(Inflation,Unemployment,Interest Rate) into a 
#single value to measure overall pressure on the economy.

df['Economic Pressure Index']= (df['Inflation Rate (%)']+df['Unemployment Rate (%)']+df['Interest Rate (%)'])/3
```

```python
df['Economic Pressure Index'].head()
```

```python
#Market Sentiment Indicator-Market sentiment reflects how optimistic or pessimistic people are about the economy and markets.
                            #Stock market is influenced by consumer confidence and GDP growth.

df['Market Sentiment Score']= (df['Consumer Confidence Index']*df['GDP Growth (%)'])
```

```python
df[['Date','Market Sentiment Score']].head()
```

```python
#Verify New Columns
df[['Date','Daily Stock Return(%)','Daily Volatility(%)','Economic Pressure Index','Market Sentiment Score']].head()
```

**Filtering Data**

```python
#Filter high GDP Growth days when GDP Growth >5%

high_gdp_days= df[df['GDP Growth (%)']>5]
high_gdp_days.head()
```

Filter by Stock Return- Days with daily stock return > 1%

```python
#Filter by Stock Return
high_Stock_Return = df[df['Daily Stock Return(%)']>1]
high_Stock_Return.head()
```

```python
#Filter by high Stock Return and high Economic Pressure

df_filtered= df[(df['Daily Stock Return(%)']>1)&(df['Economic Pressure Index']>5)]
df_filtered.head()
```

```python
#Filtering when Economic Pressure is high

high_pressure_days = df[df['Economic Pressure Index'] > 6]
high_pressure_days.head()
```

**Aggregation of Data**

```python
df['Date']= pd.to_datetime(df['Date'],format='%d-%m-%Y')
df['Year']= df['Date'].dt.year
```

```python
#Stock Market Aggregation-Average Open & Close Price

avg_prices= df[['Open Price','Close Price']].mean()
avg_prices
```

```python
#Total Trading volume

total_trading_volume= df['Trading Volume'].sum()
total_trading_volume
```

```python
#Total Trading volume per Stock Index

volume_index= df.groupby('Stock Index')['Trading Volume'].sum()
volume_index
```

```python
#Average Economic Indicators

economic_avg= df[['GDP Growth (%)','Inflation Rate (%)','Unemployment Rate (%)','Interest Rate (%)']].mean()
economic_avg
```

```python
#Yearly Summary

yearly_summary =df.groupby('Year').agg({'Close Price':'mean','Trading Volume':'sum','GDP Growth (%)':'mean','Inflation Rate (%)':'mean',
                                        'Unemployment Rate (%)':'mean','Interest Rate (%)':'mean'}).reset_index()
yearly_summary
```

**Aggregation was applied to summarize stock market and economic indicators. Average prices, total trading volume, and mean economic indicators were calculated to observe overall market behavior and economic conditions.**

Exploratory Data Analysis (EDA) is performed to understand the underlying patterns, trends, and relationships within financial and economic indicators such as GDP growth, inflation, unemployment, and stock market performance. This section includes univariate, bivariate, and multivariate analysis

**Descriptive Statistical Analysis**: 
First, descriptive statistics are used to summarize the central tendency and dispersion of numerical variables.

<p style="font-size:20px;"><b>Univariate Analysis</b><br>

Univariate analysis examines each variable independently to understand its distribution and behavior.

```python
# GDP Growth Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['GDP Growth (%)'])
plt.title('Distribution of GDP Growth')
plt.xlabel('GDP Growth (%)')
plt.ylabel('Frequency')
plt.show()
```

Insight:

The histogram shows that GDP growth rates are spread across a wide range, from negative values (economic contraction) to high positive values (economic expansion). However, most of the values are concentrated in the positive growth range, indicating that the economy experienced growth more often than decline.

```python
# Stock Market trend
plt.figure(figsize=(8,5))
sns.lineplot(x='Year', y='Stock Index', data=df)
plt.title('Stock Market Index Over Time')
plt.show()
```

Observation:

The line plot shows that the stock market index fluctuates over time rather than moving in a straight upward trend. After a decline in the early years, the index recovers and rises around the mid-period, followed by another downturn, and then shows signs of recovery toward the end. This pattern indicates that the stock market experiences cyclical movements influenced by changing economic conditions.

<p style="font-size:20px;"><b>Bivariate Analysis</b><br>

Bivariate analysis helps identify relationships between two variables.

```python
# GDP Growth vs Stock Market Performance
plt.figure(figsize=(8,5))
sns.scatterplot(x='GDP Growth (%)', y='Stock Index', data=df)
plt.title('GDP Growth vs Stock Market Index')
plt.show()
```

Insight:

The scatter plot shows no strong direct relationship between GDP growth and stock market performance, indicating that GDP growth alone does not determine stock index movements.This suggests that stock market performance is influenced by multiple factors beyond economic growth, such as interest rates, investor sentiment, and global market conditions.

```python
# Inflation vs Unemployment
plt.figure(figsize=(8,5))
sns.scatterplot(x='Inflation Rate (%)', y='Unemployment Rate (%)', data=df)
plt.title('Inflation vs Unemployment')
plt.show()
```

Observation:

This scatter plot is used to analyze the relationship between economic growth and stock market performance. A positive trend indicates that higher GDP growth is generally associated with stronger stock market performance.The inflation versus unemployment scatter plot helps identify patterns between price levels and employment.

The scatter plot shows a widely dispersed pattern with no clear linear trend between inflation and unemployment.This indicates that changes in inflation do not significantly explain variations in unemployment during the observed period.

**Correlation Analysis**

Correlation analysis is used to measure the strength and direction of the relationship between two or more variables.

In a project on financial and economic indicators, variables like GDP growth, inflation, unemployment, and stock market prices are interconnected, so correlation helps us understand how they move together.

```python
economic_vars = ['Close Price',
    'GDP Growth (%)',
    'Inflation Rate (%)',
    'Unemployment Rate (%)',
    'Interest Rate (%)',
    'Consumer Confidence Index']

corr_matrix = df[economic_vars].corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Economic Indicators')
plt.show()
```

Insight:

The correlation heatmap reveals that most financial and economic indicators exhibit weak correlations with one another. Stock close prices show negligible correlation with GDP growth, inflation, and unemployment. Similarly, inflation and interest rates do not demonstrate strong linear relationships.The absence of strong correlations suggests that stock prices and macroeconomic indicators may be influenced by multiple factors and non-linear relationships rather than simple linear dependence.

<p style="font-size:22px;"><b>Visualizations</b><br>

Visualizations are used to identify trends, patterns, and relationships among financial and economic indicators. Matplotlib and Seaborn libraries are used to create clear and informative charts with appropriate titles, labels, legends, and color schemes.

**1)Line Chart** – Stock Close Price Over Time

```python
# Stock Close Price Over Time
#Aggregate Data by Year (Mean Values)
yearly_data = df.groupby('Year')['Close Price'].mean().reset_index()
print(yearly_data)

#Plot Yearly Line Chart
plt.figure(figsize=(8,5))
plt.plot(
    yearly_data['Year'],
    yearly_data['Close Price'],
    marker='o',
    label='Yearly Avg Close Price')

plt.title("Yearly Average Stock Close Price")
plt.xlabel("Year")
plt.ylabel("Average Close Price")
plt.legend()
plt.grid(True)
plt.show()
```

Insight:

The yearly average stock close price shows fluctuations over time, with a noticeable dip in the mid-period followed by a strong rise toward the end, indicating recovery and overall growth in later years.


**2) Bar Plot** – Average GDP Growth by Stock Index

```python
#Average GDP Growth by Stock Index
avg_gdp = df.groupby('Stock Index')['GDP Growth (%)'].mean()

plt.figure(figsize=(6,4))
avg_gdp.plot(kind='bar', color='green')
plt.title("Average GDP Growth by Stock Index")
plt.xlabel("Stock Index")
plt.ylabel("GDP Growth (%)")
plt.show()
```

Insight:

The bar chart shows slight differences in average GDP growth across stock indices, with Dow Jones having the highest average growth, followed closely by NASDAQ and S&P 500, indicating broadly similar economic conditions across markets.

**3) Histogram** – Distribution of Daily Stock Returns

```python
#Distribution of Daily Stock Returns

plt.figure(figsize=(8,5))
plt.hist(df['Daily Stock Return(%)'], bins=15, color='skyblue', edgecolor='black')
plt.title("Distribution of Daily Stock Returns")
plt.xlabel("Daily Return (%)")
plt.ylabel("Frequency")
plt.show()
```

Insight:

Displays how frequently different return values occur.The histogram shows that daily stock returns are centered around zero, with most returns being small gains or losses, while extreme positive or negative returns occur less frequently, indicating typical market volatility.

**4) Box Plot** – Effect of Stock Index on Returns

```python
#Effect of Stock Index on Returns

plt.figure(figsize=(8,5))
sns.boxplot(x='Stock Index', y='Daily Stock Return(%)', data=df, palette='Set2')
plt.title("Stock Return Distribution by Index")
plt.xlabel("Stock Index")
plt.ylabel("Daily Return (%)")
plt.show()
```

Insight:

Highlights volatility and outliers across indices.
- All three indices (Dow Jones, S&P 500, and NASDAQ) have median daily returns close to zero, indicating that on most days gains and losses are relatively balanced.

- NASDAQ shows slightly higher volatility, as seen from a wider spread and more extreme outliers. This suggests NASDAQ experiences larger daily price fluctuations compared to the other indices.

- Dow Jones appears more stable, with a comparatively narrower interquartile range, indicating less variability in daily returns.

- The presence of both positive and negative outliers across all indices highlights periods of unusually high gains or losses, reflecting market sensitivity to economic and financial events.

**5) Scatter Plot** – GDP Growth vs Stock Returns

```python
#GDP Growth vs Stock Returns

plt.figure(figsize=(8,5))
sns.scatterplot(
    x=df['GDP Growth (%)'],
    y=df['Daily Stock Return(%)'],
    hue=df['Stock Index'],
    palette='deep')

plt.title("GDP Growth vs Stock Market Returns")
plt.xlabel("GDP Growth (%)")
plt.ylabel("Daily Return (%)")
plt.legend(title="Stock Index")
plt.show()
```

Insight:

Shows the relationship between economic growth and market performance. The scatter plot shows no strong correlation between GDP growth and daily stock market returns for any of the indices, indicating that short-term stock returns are largely independent of GDP growth.The points are widely scattered, suggesting that GDP growth alone does not strongly determine short-term stock returns. 

**6) Subplots** – Multiple Economic Indicators

```python
# Multiple Economic Indicators

fig, axes = plt.subplots(2, 2, figsize=(12,8))

sns.lineplot(ax=axes[0,0], x=df.index, y=df['GDP Growth (%)'], color='green')
axes[0,0].set_title("GDP Growth Trend")

sns.lineplot(ax=axes[0,1], x=df.index, y=df['Inflation Rate (%)'], color='red')
axes[0,1].set_title("Inflation Rate Trend")

sns.lineplot(ax=axes[1,0], x=df.index, y=df['Unemployment Rate (%)'], color='purple')
axes[1,0].set_title("Unemployment Rate Trend")
sns.lineplot(ax=axes[1,1], x=df.index, y=df['Interest Rate (%)'], color='blue')
axes[1,1].set_title("Interest Rate Trend")

plt.tight_layout()
plt.show()
```

Insight:

Allows comparison of multiple economic indicators in one view.The plots show that GDP growth, inflation, unemployment, and interest rates fluctuate significantly over time, indicating frequent changes in macroeconomic conditions rather than a stable long-term trend.
